```python
import pandas as pd
from sklearn.svm import LinearSVC
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

### Simple classifier using Linear SVM 

```python
def classify_dataset(train_file, test_file, dataset_name):
    print("\n======================================")
    print(dataset_name)
    print("======================================")

    # Load data
    train_df = pd.read_csv(train_file)
    test_df = pd.read_csv(test_file)

    print("Training pattern shape:", train_df.shape)
    print("Test pattern shape:", test_df.shape)

    # Split features and labels
    X_train = train_df.drop(columns=["label"])
    y_train = train_df["label"]

    X_test = test_df.drop(columns=["label"])
    y_test = test_df["label"]

    # Normalize features
    scaler = MinMaxScaler()
    X_train = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)

    # Train Linear SVM
    classifier = LinearSVC(max_iter=1000)
    classifier.fit(X_train, y_train)

    # Calculate accuracy
    train_predictions = classifier.predict(X_train)
    test_predictions = classifier.predict(X_test)

    train_accuracy = accuracy_score(y_train, train_predictions) * 100
    test_accuracy = accuracy_score(y_test, test_predictions) * 100

    print("Number of features:", X_train.shape[1])
    print("Training Accuracy:", round(train_accuracy, 2), "%")
    print("Test Accuracy:", round(test_accuracy, 2), "%")

    # Evaluation
    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, test_predictions))

    print("\nClassification Report:")
    print(classification_report(y_test, test_predictions))

    return train_accuracy, test_accuracy
```

```python
# ======================================================
# FEI 1
# ======================================================

f1_train_accuracy, f1_test_accuracy = classify_dataset(
    "../data/f1/f1_train_patterns.csv",
    "../data/f1/f1_test_patterns.csv",
    "FEI 1"
)


# ======================================================
# FEI 2
# ======================================================

f2_train_accuracy, f2_test_accuracy = classify_dataset(
    "../data/f2/f2_train_patterns.csv",
    "../data/f2/f2_test_patterns.csv",
    "FEI 2"
)


# ======================================================
# Final Results
# ======================================================

print("\n======================================")
print("FINAL RESULTS")
print("======================================")

print(
    "FEI 1 - Training Accuracy:",
    round(f1_train_accuracy, 2), "%"
)

print(
    "FEI 1 - Test Accuracy:",
    round(f1_test_accuracy, 2), "%"
)

print(
    "FEI 2 - Training Accuracy:",
    round(f2_train_accuracy, 2), "%"
)

print(
    "FEI 2 - Test Accuracy:",
    round(f2_test_accuracy, 2), "%"
)
```
